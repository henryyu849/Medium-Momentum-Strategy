```python
def strategy(k):
    for j in range(mediummomentum.shape[0]): # for all rows
        portfolio_returns = []
        for i in range(k): # where k is the holding period 
            if j-k+i < 0: #if we are in the first k months, we cannot hold a portfolio for k months, so we will skip this iteration
                continue
            monthportfolioreturns = []
            mediummomentum["top_decile"].iloc[j-k+i] = top_list #we have extracted the top decile k months ago, and we want to hold it for k months, 
            for a in range(len(top_list)):
                stockreturns = 1
                for b in range(i+1):
                    list_number = top_list[a]
                    stockreturns *= (1 + returns_m.iloc[j-b, list_number]) #we want to add the position of the stock in the returns_m dataframe to the list of stocks in the top decile, so that we can use it to calculate the returns later on.
                stockreturns = stockreturns - 1
                monthportfolioreturns.append(stockreturns) 
            averagereturn = sum(monthportfolioreturns)/len(monthportfolioreturns)
            portfolio_returns.append(averagereturn)
        weighted_average = sum(portfolio_returns)/len(portfolio_returns)
        mediummomentum["portfolio_return"].iloc[j] = weighted_average
```

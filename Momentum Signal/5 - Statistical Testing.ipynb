```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import wilcoxon

r = portfolio_returns["1 Month"].dropna()

T = len(r)
lags = 0

X = np.ones(T)                                    # constant only
res = sm.OLS(r.values, X).fit(cov_type="HAC", cov_kwds={"maxlags": lags}) # fit the OLS model with Newey-West standard errors as the residuals are likely to be autocorrelated and heteroskedastic.

print(res.summary())
print("mean:", res.params[0])
print("NW t-stat:", res.tvalues[0])
print("p-value:", res.pvalues[0])

from scipy.stats import wilcoxon

stat, p = wilcoxon(r, alternative="two-sided") # perform the Wilcoxon signed-rank test to determine if the median of the returns is significantly different from zero. This is a non-parametric test that does not assume normality of the data.
print(stat, p)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import wilcoxon

K_map = {"1 Month": 1, "3 Months": 3, "6 Months": 6, "12 Months": 12}   

common = portfolio_returns[list(K_map)].dropna().index
pr = portfolio_returns.loc[common]

rows = []
for col, K in K_map.items():
    r = pr[col].dropna()
    res = sm.OLS(r.values, np.ones(len(r))).fit(
        cov_type="HAC", cov_kwds={"maxlags": max(K - 1, 1)})
    rows.append({
        "K": K,
        "N": len(r),
        "mean_monthly": r.mean(),
        "NW_t": res.tvalues[0],
        "NW_p": res.pvalues[0],
        "wilcoxon_p": wilcoxon(r).pvalue,
        "ann_sharpe": r.mean() / r.std() * np.sqrt(12),
        "skew": r.skew(),
        "min": r.min(),
    })

summary = pd.DataFrame(rows).set_index("K") # create a summary dataframe with the results of the statistical tests and descriptive statistics for each holding period K.
print(summary.round(4))
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

r = threemonthhold["portfolio_return"].dropna()

T = len(r)
lags = 2

X = np.ones(T)                                    # constant only
res = sm.OLS(r.values, X).fit(cov_type="HAC", cov_kwds={"maxlags": lags})

print(res.summary())
print("mean:", res.params[0])
print("NW t-stat:", res.tvalues[0])
print("p-value:", res.pvalues[0])
```

```python
from scipy.stats import wilcoxon

stat, p = wilcoxon(r, alternative="two-sided")
print(stat, p)
```

```python
mediummomentum = pd.DataFrame(index=returns_m.index, columns=[f'12M Momentum for {col}' for col in returns_m.columns])

for k in range(returns_m.shape[1]):
    for j in range(returns_m.shape[0]):
        if j < 12:
            mediummomentum.iloc[j, k] = None # if there are less than 12 months of data, we cannot calculate the momentum, so we set it to None.
        else:
            momentum = 1
            for i in range(j-12, j-1):
                formation = 1+returns_m.iloc[i,k]
                momentum = momentum * formation # calculate the momentum by multiplying the returns for the past 12 months.
            momentum = momentum - 1
            mediummomentum.iloc[j, k] = momentum  # store the momentum in the mediummomentum dataframe.

print(mediummomentum.tail())
```

```python
mediummomentum = pd.DataFrame(index=returns_m.index, columns=[f'12M Momentum for {col}' for col in returns_m.columns])

for k in range(returns_m.shape[1]):
    for j in range(returns_m.shape[0]):
        if j < 12:
            mediummomentum.iloc[j, k] = None
        else:
            momentum = 1
            for i in range(j-12, j-1):
                formation = 1+returns_m.iloc[i,k]
                momentum = momentum * formation
            momentum = momentum - 1
            mediummomentum.iloc[j, k] = momentum

print(mediummomentum.tail())
```
